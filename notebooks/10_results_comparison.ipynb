```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

results = pd.DataFrame({
    "Model": [
        "Naive",
        "Mean",
        "MA20",
        "Momentum",
        "LightGBM",
        "LSTM",
        "Stacking"
    ],
    "RMSE": [
        0.01602,
        0.01600,
        0.01650,
        0.02350,
        0.01590,
        0.01615,
        0.01585
    ],
    "Direction_Acc": [
        0.50,
        0.5407,
        0.5407,
        0.5407,
        0.53,
        0.4846,
        0.5458
    ],
    "Sharpe": [
        np.nan,
        np.nan,
        np.nan,
        np.nan,
        np.nan,
        np.nan,
        -0.52
    ]
})

print("FINAL MODEL COMPARISON")
print(results)

plt.figure(figsize=(10,5))
plt.bar(results["Model"], results["Direction_Acc"])
plt.xticks(rotation=45)
plt.title("Direction Accuracy Comparison")
plt.show()

plt.figure(figsize=(10,5))
plt.bar(results["Model"], results["RMSE"])
plt.xticks(rotation=45)
plt.title("RMSE Comparison")
plt.show()

best_model = results.loc[results["Direction_Acc"].idxmax(), "Model"]

print("\nKEY INSIGHT")
print("Best model by Direction Accuracy:", best_model)

print("\nNOTE:")
print("Even best ML models only slightly outperform baselines")
print("Financial predictability is extremely limited at daily frequency")
```
